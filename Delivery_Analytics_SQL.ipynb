# Delivery Analytics com SQL

Esse é um projeto de análise de dados com SQL (PostgreSQL) usando o banco de dados fictício de um app de delivery.

| Tabela | Papel | O que tem lá |
|---|---|---|
| `pedidos` | Fato | 1 linha por pedido: valor, status, tempo de entrega |
| `clientes` | Dimensão | Quem faz o pedido (cidade, idade) |
| `restaurantes` | Dimensão | Quem vende (categoria, taxa de entrega) |
| `entregadores` | Dimensão | Quem entrega (veículo, cidade) |
| `itens_pedido` | Detalhe | Os produtos de cada pedido (relação 1:N com pedidos) |
| `avaliacoes` | Feedback | Nota do cliente (deixei incompleto de propósito, pois nem todo mundo avalia) |

## Olhando os dados pela primeira vez

Antes de mergulhar nas perguntas mais complexas, dei uma olhada na distribuição dos status dos pedidos para entender o cenário:

| Status | Quantidade | % do total |
|---|---|---|
| Entregue | 2.031 | 81,2% |
| Em andamento | 247 | 9,9% |
| Cancelado | 222 | 8,9% |

Como mais de 80% dos pedidos foram entregues, temos um volume bom para analisar a receita sem que os cancelamentos atrapalhem muito a leitura dos números.

## Respondendo perguntas de negócio

### Qual categoria de restaurante gera mais dinheiro?

Fiz um `JOIN` entre as tabelas de `pedidos` e `restaurantes` para ver de onde vem a receita. Dos mais de R$ 183 mil faturados, a distribuição ficou assim:

| Categoria | Receita |
|---|---|
| Lanchonete | R$ 34.462,84 |
| Pizzaria | R$ 29.210,65 |
| Brasileira | R$ 24.491,40 |
| Mexicana | R$ 19.140,92 |
| Doceria | R$ 18.704,61 |
| Italiana | R$ 16.633,61 |
| Árabe | R$ 12.642,03 |
| Japonesa | R$ 11.607,61 |
| Hamburgueria | R$ 11.164,67 |
| Vegetariana | R$ 4.971,33 |

**Lanchonete disparou na frente**, gerando quase 7 vezes mais que a categoria Vegetariana. Mas fica a curiosidade: será que eles vendem muito mais em quantidade ou o ticket médio é que é mais alto? Essa seria uma ótima pergunta para uma parte 2 do projeto.

```sql
                                                                         SQL: Qual categoria de restaurante gera mais dinheiro?
SELECT restaurantes.categoria, SUM(pedidos.valor_total)
FROM pedidos
JOIN restaurantes ON pedidos.restaurante_id = restaurantes.restaurante_id
WHERE pedidos.status = 'entregue'
GROUP BY restaurantes.categoria
ORDER BY SUM(pedidos.valor_total) DESC;
```

### O meio de pagamento afeta o valor do pedido?

| Forma de pagamento | Ticket médio |
|---|---|
| Cartão | R$ 93,14 |
| Vale-refeição | R$ 90,17 |
| Dinheiro | R$ 89,18 |
| Pix | R$ 87,86 |

Rodando a média de valor, vi que os números são bem parecidos (variam menos de R$ 6). Ou seja, não parece que o cliente gasta mais ou menos dependendo se paga no Pix ou no cartão.

### Qual a porcentagem de clientes que realmente avalia o pedido?

- Pedidos entregues: **2.031**
- Pedidos entregues **com** avaliação: **1.411**
- Taxa de avaliação: **69,5%**

```sql
                                                                              SQL: Porcentagem de clientes que avalia o pedido
SELECT COUNT(*)
FROM pedidos
LEFT JOIN avaliacoes ON pedidos.pedido_id = avaliacoes.pedido_id
WHERE pedidos.status = 'entregue' AND avaliacoes.nota IS NOT NULL;
```

## Considerações finais

Mesmo usando queries mais iniciais (agrupamentos, filtros e joins), deu para tirar insights muito legais de negócio. Identificamos qual tipo de restaurante lidera as vendas e levantamos pontos de atenção importantes, como a lacuna nas avaliações.

Como os dados são sintéticos e foram gerados por mim de forma aleatória, os resultados servem mais para demonstrar o domínio do SQL e o raciocínio por trás da análise do que para tirar conclusões reais de mercado. De qualquer forma, foi um excelente exercício prático para o meu portfólio.
