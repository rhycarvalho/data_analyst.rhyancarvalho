# SQL  Banco `biblioteca`

### 1. Título e ano de publicação de todos os livros


```python

SELECT titulo, ano_publicacao
FROM livros
ORDER BY titulo;
```

### 2. Livros de Romance, do mais antigo para o mais novo


```python

SELECT titulo, ano_publicacao
FROM livros
WHERE genero = 'Romance'
ORDER BY ano_publicacao;
```

### 3. Membros mais novos da base (cadastrados a partir de 01/06/2023)


```python

SELECT nome, data_cadastro
FROM membros
WHERE data_cadastro >= '2023-06-01'
ORDER BY data_cadastro;
```

### 4. Quantos livros estão fora da biblioteca agora



```python

SELECT COUNT(*) AS emprestimos_em_aberto
FROM emprestimos
WHERE data_devolucao_real IS NULL;
```

### 5. Título de cada livro junto com o nome do autor


```python

SELECT l.titulo, a.nome AS autor
FROM livros l
JOIN autores a ON a.id_autor = l.id_autor
ORDER BY l.titulo;
```

### 6. Quantos livros cada autor tem no acervo


```python

SELECT a.nome, COUNT(l.id_livro) AS total_livros
FROM autores a
LEFT JOIN livros l ON l.id_autor = a.id_autor
GROUP BY a.nome
ORDER BY total_livros DESC;
```

### 7. Gênero com mais títulos no acervo


```python

SELECT genero, COUNT(*) AS total
FROM livros
GROUP BY genero
ORDER BY total DESC
LIMIT 1;
```

### 8. Valor total do acervo, por gênero

```python

SELECT genero, SUM(preco) AS valor_total
FROM livros
GROUP BY genero
ORDER BY valor_total DESC;
```

### 9. Empréstimos devolvidos com atraso

Compara duas datas da mesma linha: a prevista e a real. Só entra na contagem quem já devolveu de fato (`IS NOT NULL`) — quem ainda está com o livro não conta como atrasado nem em dia, porque ainda não sabemos.

```python

SELECT COUNT(*) AS emprestimos_atrasados
FROM emprestimos
WHERE data_devolucao_real IS NOT NULL
  AND data_devolucao_real > data_devolucao_prevista;
```
