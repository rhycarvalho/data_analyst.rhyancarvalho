
Essas primeiras eu resolvi só com `SELECT`, `WHERE`, `ORDER BY`

### 1. Usuários do Rio de Janeiro

Listar nome e cidade dos usuários que moram no Rio de Janeiro.

```python

SELECT nome, cidade
FROM usuarios
WHERE cidade = 'Rio de Janeiro'

```

### 2. Quantos usuários ativos


```python

SELECT COUNT(*) AS total_ativos
FROM usuarios
WHERE ativo
```

### 3. Séries lançadas a partir de 2023


```python

SELECT nome, genero, ano_lancamento
FROM titulos
WHERE tipo = 'Série' AND ano_lancamento >= 2023
ORDER BY ano_lancamento DESC
```

### 4. Preço médio dos planos


```python

SELECT AVG(preco_mensal) AS media_precos
FROM planos

```

### 5. Quantas avaliações com nota 5


```python

SELECT COUNT(nota) AS total_nota_5
FROM avaliacoes
WHERE nota = 5

```

### 6. Nome do usuário + nome do plano nas assinaturas ativas

```python

SELECT
    usuarios.nome AS nome_usuario,
    planos.nome_plano
FROM assinaturas
JOIN usuarios ON assinaturas.id_usuario = usuarios.id_usuario
JOIN planos ON assinaturas.id_plano = planos.id_plano
WHERE assinaturas.status = 'Ativa'

```

### 7. Quantidade de títulos por gênero


```python

SELECT genero, COUNT(*) AS total_titulos
FROM titulos
GROUP BY genero
ORDER BY total_titulos DESC
```

### 8. Soma dos pagamentos aprovados

```python
SELECT SUM(valor) AS total
FROM pagamentos
WHERE status_pagamento = 'Aprovado'

```

### 9. Os 5 usuários com cadastro mais antigo


```python

SELECT nome, data_cadastro
FROM usuarios
ORDER BY data_cadastro
LIMIT 5

```

### 10. Tempo médio assistido por dispositivo


```python

SELECT dispositivo, AVG(tempo_assistido_min) AS tempo_medio_min
FROM historico_visualizacao
GROUP BY dispositivo
```
